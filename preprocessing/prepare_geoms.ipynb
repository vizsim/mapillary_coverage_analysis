```python
import geopandas as gpd

# gdf = gpd.read_file("data/DE_VG5000.gpkg")
# gdf
```

```python
# Layer-Übersicht: alle Layer mit Geometrietyp
gpkg_path = "data/DE_VG5000.gpkg"
layer_overview = gpd.list_layers(gpkg_path)
layer_overview
```

```python
# Nur Layer mit Polygon oder Multi Polygon
poly_layers = layer_overview[
    layer_overview["geometry_type"].astype(str).str.upper().str.replace(" ", "").isin(
        ["POLYGON", "MULTIPOLYGON"]
    )
].copy()
poly_layers
```

```python
# Alle Polygon/Multipolygon-Layer laden und kurz untersuchen (Spalten, CRS, Stichprobe)
for layer_name in poly_layers["name"]:
    gdf_layer = gpd.read_file(gpkg_path, layer=layer_name)
    print(f"=== {layer_name} ===\n  Zeilen: {len(gdf_layer)}, CRS: {gdf_layer.crs}, Spalten: {list(gdf_layer.columns)}\n")
```

```python
# Die 3 gewünschten Layer einlesen: Länder, Kreise, Gemeinden
gdf_lan = gpd.read_file(gpkg_path, layer="vg5000_lan")
gdf_krs = gpd.read_file(gpkg_path, layer="vg5000_krs")
gdf_gem = gpd.read_file(gpkg_path, layer="vg5000_gem")

gdf_lan, gdf_krs, gdf_gem
```

```python
# Die 3 Layer als GeoParquet in data/ schreiben (je eine .parquet-Datei pro Layer)
gdf_lan.to_parquet("data/bkg/lan.parquet")
gdf_krs.to_parquet("data/bkg/krs.parquet")
gdf_gem.to_parquet("data/bkg/gem.parquet")
print("Geschrieben: data/bkg/lan.parquet, data/bkg/krs.parquet, data/bkg/gem.parquet")
```
