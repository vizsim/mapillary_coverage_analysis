### BKG als einzige Quelle (3 Ebenen)

- **Bundesland** (lan), **Kreis** (krs), **Gemeinde** (gem) kommen nur noch aus dem VG5000-GPKG unter `preprocessing/data/bkg/`.
- Ein räumlicher Join erfolgt einmal gegen die View `v_vz5000_gem` (enthält GEN_L, GEN_K, GEN_G) → jede Straße bekommt alle drei Zuordnungen.
- Spaltenname im GDF: **Kreis** (BKG spricht von "Kreis", nicht "Landkreis"); Aggregation über `r_einheit` wählbar.

```python
# BKG VG5000: 3 Ebenen aus einer Quelle (preprocessing/data/bkg/DE_VG5000.gpkg)
# Für sjoin: eine Ebene mit Hierarchie (v_vz5000_gem) → Bundesland, Kreis, Gemeinde in einem Schritt
gem_hierarchy = gpd.read_file(BKG_GPKG, layer=BKG_LAYER_GEM_HIERARCHY)
gem_hierarchy = gem_hierarchy.rename(columns={"GEN_L": "Bundesland", "GEN_K": "Kreis", "GEN_G": "Gemeinde"})
# AGS_0 (Amtlicher Gemeindeschlüssel) für eindeutige Zuordnung (mehrere Gemeinden können gleichen Namen haben)
if "AGS_G" in gem_hierarchy.columns:
    gem_hierarchy["AGS_0"] = gem_hierarchy["AGS_G"]

# Flächen für Karten-Export (PMTiles/FGB) pro Ebene
gdf_lan = gpd.read_file(BKG_GPKG, layer=BKG_LAYER_LAN).rename(columns={"GEN": "Bundesland"})
gdf_krs = gpd.read_file(BKG_GPKG, layer=BKG_LAYER_KRS).rename(columns={"GEN": "Kreis"})
gdf_gem = gpd.read_file(BKG_GPKG, layer=BKG_LAYER_GEM).rename(columns={"GEN": "Gemeinde"})
```

```python
pts = lines[["osm_id", "geometry"]].copy()
pts["geometry"] = pts.geometry.representative_point()
```

```python
hierarchy_cols = [c for c in ["Bundesland", "Kreis", "Gemeinde", "AGS_0", "geometry"] if c in gem_hierarchy.columns]
joined = pts.sjoin(
    gem_hierarchy[hierarchy_cols],
    how="left",
    predicate="within",
)
```

```python
join_cols = [c for c in ["osm_id", "Bundesland", "Kreis", "Gemeinde", "AGS_0"] if c in joined.columns]
lines = lines.merge(
    joined[join_cols],
    on="osm_id",
    how="left",
)
```

```python
lines["length_m"] = lines.geometry.length
```

```python

df = lines.copy()

# NaN explizit
df["mapillary_coverage"] = df["mapillary_coverage"].fillna("NaN")

# _link entfernen → motorway_link → motorway
df["highway"] = df["highway"].str.replace("_link$", "", regex=True)
```

```python
##### Aggregation: gleiche Pipeline pro Ebene (Bundesland, Kreis, Gemeinde) im Loop
```

```python
EBENEN = ["Bundesland", "Kreis", "Gemeinde"]
wide_by_ebene = {}
cov_order = ["NaN", "pano", "regular"]
order = ["motorway", "trunk", "primary", "secondary", "tertiary", "all"]

for r_einheit in EBENEN:
    # Bei Gemeinden nach AGS_0 gruppieren (eindeutig), nicht nach Name (mehrere "Stetten" etc.)
    group_key = "AGS_0" if r_einheit == "Gemeinde" else r_einheit
    agg = (
        df.groupby([group_key, "highway", "mapillary_coverage"], observed=True)["length_m"]
        .sum()
        .reset_index()
    )
    agg["total_length"] = agg.groupby([group_key, "highway"])["length_m"].transform("sum")
    agg["share"] = agg["length_m"] / agg["total_length"]

    agg_all = (
        df.groupby([group_key, "mapillary_coverage"], observed=True)["length_m"]
        .sum()
        .reset_index()
    )
    agg_all["total_length"] = agg_all.groupby(group_key)["length_m"].transform("sum")
    agg_all["share"] = agg_all["length_m"] / agg_all["total_length"]
    agg_all["highway"] = "all"

    agg_combined = pd.concat([agg, agg_all], ignore_index=True)
    _pivot_df = agg_combined.copy()
    _pivot_df["mapillary_coverage"] = _pivot_df["mapillary_coverage"].astype("string").fillna("NaN")

    share_wide = (
        _pivot_df.pivot_table(
            index=[group_key, "highway"],
            columns="mapillary_coverage",
            values="share",
            aggfunc="sum",
            fill_value=0,
        )
    )
    length_wide = (
        _pivot_df.pivot_table(
            index=[group_key, "highway"],
            columns="mapillary_coverage",
            values="length_m",
            aggfunc="sum",
            fill_value=0,
        )
        .div(1000.0)
        .round(1)
    )
    for c in cov_order:
        if c not in share_wide.columns:
            share_wide[c] = 0
        if c not in length_wide.columns:
            length_wide[c] = 0
    share_wide = share_wide[cov_order].add_prefix("share_")
    length_wide = length_wide[cov_order].add_prefix("length_")
    result = pd.concat([share_wide, length_wide], axis=1).reset_index()

    result["highway"] = pd.Categorical(result["highway"], order, ordered=True)
    result = result.sort_values([group_key, "highway"])

    r = result.copy().rename(columns={
        "share_NaN": "share_no_cover",
        "length_NaN": "length_no_cover",
    })
    wide_long = r.melt(
        id_vars=[group_key, "highway"],
        value_vars=[
            "share_no_cover", "share_pano", "share_regular",
            "length_no_cover", "length_pano", "length_regular",
        ],
        var_name="metric_coverage",
        value_name="value",
    )
    wide_long[["metric", "coverage"]] = wide_long["metric_coverage"].str.split("_", n=1, expand=True)
    wide_long["col"] = (
        wide_long["highway"].astype("string")
        + "_" + wide_long["metric"].astype("string")
        + "_" + wide_long["coverage"].astype("string")
    )
    wide = (
        wide_long.pivot(index=group_key, columns="col", values="value")
        .fillna(0)
        .reset_index()
    )
    wide_by_ebene[r_einheit] = wide

wide_bundesland = wide_by_ebene["Bundesland"]
wide_kreis = wide_by_ebene["Kreis"]
wide_gemeinde = wide_by_ebene["Gemeinde"]
# Gemeindenamen für Anzeige (AGS_0 ist der eindeutige Schlüssel)
if "AGS_0" in wide_gemeinde.columns:
    wide_gemeinde = wide_gemeinde.merge(
        gdf_gem[["AGS_0", "Gemeinde"]].drop_duplicates(), on="AGS_0", how="left"
    )
wide = wide_bundesland  # Rückwärtskompatibilität / Standard für Anzeige 
```

```python
# ...existing code...
def fgb_to_pmtiles_layer(polys_fgb, output_pmtiles, layer_name, minzoom, maxzoom, include_fields=None):
    import subprocess
    from pathlib import Path

    cmd = [
        "tippecanoe",
        "-o", str(Path(output_pmtiles).resolve()),
        f"--layer={layer_name}",
        f"--minimum-zoom={minzoom}",
        f"--maximum-zoom={maxzoom}",
        "--force",
        "--read-parallel",
        "--detect-shared-borders",
        "--coalesce-densest-as-needed",
        "--extend-zooms-if-still-dropping",
    ]

    if include_fields:
        for f in include_fields:
            cmd += ["-y", f]

    cmd.append(str(Path(polys_fgb).resolve()))
    subprocess.run(cmd, check=True)

    print(f"✅ PMTiles created: {output_pmtiles}")
# ...existing code...
```
