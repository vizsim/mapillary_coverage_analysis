```python
import geopandas as gpd
import pandas as pd
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor
import time
```

```python
gdf_all=gpd.read_parquet("data/osm/all_highways.parquet")    
```

```python
coverage_data = pd.read_csv("data/germany_osm-highways_mp-coverage_latest.csv")
```

```python
coverage_data.head()
```

```python
# prepare merge

gdf_all.osm_id = gdf_all.osm_id.astype(int)
coverage_data.osm_id = coverage_data.osm_id.astype(int)
```

```python
# merge
merged_gdf = gdf_all.merge(coverage_data, on="osm_id", how="left")
```

```python
merged_gdf
```

```python
merged_gdf_clean=merged_gdf[["osm_id", "highway", "region", "mapillary_coverage", "geometry"]].copy()
```

```python
#merged_gdf_clean.highway.unique()
```

```python
merged_gdf_clean_majorRoads=merged_gdf_clean[merged_gdf_clean.highway.isin(['motorway', 'motorway_link',
                                                'trunk', 'trunk_link', 
                                                "primary", "primary_link", 
                                                "secondary", "secondary_link", 
                                                "tertiary", "tertiary_link", 
                                                ])].copy()
```

```python
merged_gdf_clean_majorRoads
```

```python
kreise=gpd.read_file("https://raw.githubusercontent.com/isellsoap/deutschlandGeoJSON/refs/heads/main/4_kreise/1_sehr_hoch.geo.json")
bland=gpd.read_file("https://raw.githubusercontent.com/isellsoap/deutschlandGeoJSON/refs/heads/main/2_bundeslaender/1_sehr_hoch.geo.json")
```

```python
#kreise
#bland
```

```python
kreise.rename(columns={"NAME_1":"Bundesland","NAME_3": "Landkreis"}, inplace=True)
bland.rename(columns={"name":"Bundesland"}, inplace=True)
```

```python
target_crs = 25832
lines = merged_gdf_clean_majorRoads.to_crs(target_crs)
kreise  = kreise.to_crs(target_crs)  # kreise: Polygon-GDF mit Spalte "Landkreis"
```

```python
pts = lines[["osm_id", "geometry"]].copy()
pts["geometry"] = pts.geometry.representative_point()
```

```python
joined = pts.sjoin(
    kreise[["Landkreis","Bundesland", "geometry"]],
    how="left",
    predicate="within",
)
```

```python
lines = lines.merge(
    joined[["osm_id", "Landkreis","Bundesland"]],
    on="osm_id",
    how="left",
)
```

```python
lines["length_m"] = lines.geometry.length
```

```python
lines
```

```python

df = lines.copy()

# NaN explizit
df["mapillary_coverage"] = df["mapillary_coverage"].fillna("NaN")

# _link entfernen → motorway_link → motorway
df["highway"] = df["highway"].str.replace("_link$", "", regex=True)
```

```python
df
```

```python
##### AB Hier muss man sich entscheiden, ob man die Daten auf Kreisebene oder Bundeslandebene analysieren möchte.
```

```python
#r_einheit="Landkreis" 
r_einheit="Bundesland" 
```

```python
agg = (
    df.groupby([r_einheit, "highway", "mapillary_coverage"], observed=True)["length_m"]
      .sum()
      .reset_index()
)

agg["total_length"] = (
    agg.groupby([r_einheit, "highway"])["length_m"]
       .transform("sum")
)

agg["share"] = agg["length_m"] / agg["total_length"]
```

```python
agg
```

```python
agg_all = (
    df.groupby([r_einheit, "mapillary_coverage"], observed=True)["length_m"]
      .sum()
      .reset_index()
)

agg_all["total_length"] = (
    agg_all.groupby(r_einheit)["length_m"]
           .transform("sum")
)

agg_all["share"] = agg_all["length_m"] / agg_all["total_length"]

# highway-Spalte ergänzen
agg_all["highway"] = "all"
```

```python
#agg_all
```

```python
agg_combined = pd.concat([agg, agg_all], ignore_index=True)
```

```python
agg_combined
```

```python
# result = (
#     agg_combined
#       .pivot(
#           index=[r_einheit, "highway"],
#           columns="mapillary_coverage",
#           values="share"
#       )
#       .fillna(0)
#       .reset_index()
# )


cov_order = ["NaN", "pano", "regular"]

df = agg_combined.copy()

# NaN in mapillary_coverage als Kategorie behandeln
df["mapillary_coverage"] = df["mapillary_coverage"].astype("string").fillna("NaN")

# 1) Shares
share_wide = (
    df.pivot_table(
        index=[r_einheit, "highway"],
        columns="mapillary_coverage",
        values="share",
        aggfunc="sum",
        fill_value=0,
    )
)

# 2) Längen in km (auf 1 Nachkommastelle runden)
length_wide = (
    df.pivot_table(
        index=[r_einheit, "highway"],
        columns="mapillary_coverage",
        values="length_m",
        aggfunc="sum",
        fill_value=0,
    )
    .div(1000.0)
    .round(1)
)

# fehlende Spalten absichern + Reihenfolge
for c in cov_order:
    if c not in share_wide.columns:
        share_wide[c] = 0
    if c not in length_wide.columns:
        length_wide[c] = 0

share_wide = share_wide[cov_order].add_prefix("share_")
length_wide = length_wide[cov_order].add_prefix("length_")

result = pd.concat([share_wide, length_wide], axis=1).reset_index()
```

```python
result
```

```python
order = ["motorway", "trunk", "primary", "secondary", "tertiary", "all"]
result["highway"] = pd.Categorical(result["highway"], order, ordered=True)
result = result.sort_values([r_einheit, "highway"])
```

```python
result
```

```python
r = result.copy().rename(columns={
    "share_NaN": "share_no_cover",
    "length_NaN": "length_no_cover",
})

wide_long = r.melt(
    id_vars=[r_einheit, "highway"],
    value_vars=[
        "share_no_cover", "share_pano", "share_regular",
        "length_no_cover", "length_pano", "length_regular",
    ],
    var_name="metric_coverage",
    value_name="value",
)

wide_long[["metric", "coverage"]] = wide_long["metric_coverage"].str.split("_", n=1, expand=True)

# FIX: Categorical -> string
wide_long["col"] = (
    wide_long["highway"].astype("string")
    + "_" + wide_long["metric"].astype("string")
    + "_" + wide_long["coverage"].astype("string")
)

wide = (
    wide_long.pivot(index=r_einheit, columns="col", values="value")
             .fillna(0)
             .reset_index()
)

wide.head()
```

```python
# ...existing code...
def fgb_to_pmtiles_layer(polys_fgb, output_pmtiles, layer_name, minzoom, maxzoom, include_fields=None):
    import subprocess
    from pathlib import Path

    cmd = [
        "tippecanoe",
        "-o", str(Path(output_pmtiles).resolve()),
        f"--layer={layer_name}",
        f"--minimum-zoom={minzoom}",
        f"--maximum-zoom={maxzoom}",
        "--force",
        "--read-parallel",
        "--detect-shared-borders",
        "--coalesce-densest-as-needed",
        "--extend-zooms-if-still-dropping",
    ]

    if include_fields:
        for f in include_fields:
            cmd += ["-y", f]

    cmd.append(str(Path(polys_fgb).resolve()))
    subprocess.run(cmd, check=True)

    print(f"✅ PMTiles created: {output_pmtiles}")
# ...existing code...
```
